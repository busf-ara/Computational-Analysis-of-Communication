loading the rds into panda dataframe

```python
import pyreadr
import pandas as pd
from IPython.display import display
```

load data into python

```python
print("loading rds")
res = pyreadr.read_r("all_multi_paragraphs_2022_2026.RDS")
df = res[None]
```

clean up up unicode stuff

```python
print("Cleaning")
df["text_block"] = df["text_block"].astype(str).str.replace(
    r"\r\n|\r|\u2028|\u2029", "\n", regex=True
)
```

check the dataframe

```python
display(df.head(3))
```

Sample 5000 paragraphs

```python

sample_df = df.sample(n=5000, random_state=123)
```

Save

```python
sample_df.to_json("sample_for_llm.jsonl", orient="records", lines=True, force_ascii=False)
```
